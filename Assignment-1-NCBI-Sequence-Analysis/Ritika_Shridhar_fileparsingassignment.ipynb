```python
!pip install biopython

from Bio import Entrez, SeqIO
from Bio.SeqUtils import gc_fraction
from collections import Counter

Entrez.email = "rushridhar07@gmail.com"

handle = Entrez.efetch(
    db="nucleotide",
    id="NM_000546",
    rettype="fasta",
    retmode="text"
)

with open("tp53.fasta", "w") as file:
    file.write(handle.read())

handle.close()

record = SeqIO.read("tp53.fasta", "fasta")
sequence = record.seq

length = len(sequence)

A = sequence.count("A")
T = sequence.count("T")
G = sequence.count("G")
C = sequence.count("C")

gc_content = gc_fraction(sequence)*100

print("Sequence ID:", record.id)
print("Total Sequence Length:", length)
print("\nNucleotide Composition:")
print("A:", A)
print("T:", T)
print("G:", G)
print("C:", C)

print(f"\nGC Content: {gc_content:.2f}%")

print("\nFirst 30 nucleotides:")
print(sequence[:30])

print("\nLast 30 nucleotides:")
print(sequence[-30:])
```

#Written Analysis
1. The GC content i got after running the code was 53.38% which is higher than 41%. The G and C nucleotide bases form 3 hydrongen bonds instead of 2 and therefore, higher gc conetnt means higher stability of mRNA. The codons that have gc in them are less likely to break down quickly. This results in "codon usage bias". A more stable mRNA can remain in the cell for a longer period of time, hence increasing the expression level of TP53.

2. The first 30 nucleotides of the TP53 mRNA are CTCAAAAGTCTAGAGCCACCGTCCAGGGAG. As discussed earlier, A and T nucleotides form 2 hydrogen bonds which results in les stable secondary structures. The 5' UTR of mRNA being  AT-rich makes it easier for the ribosome to bind and scan it, leading to efficient transalation initiation. In this case the first 30 nucleotides have relatively high gc content especially in the second half of the sequence. Since TP53 is a tumore suppressor gene, this GC content helps with the regulation to ensure that the protein is produced at appropriate levels rather than continuously.

```python

!pip install biopython

from Bio import Entrez, SeqIO

Entrez.email = "rushridhar07@gmail.com"

handle = Entrez.efetch(
    db="nucleotide",
    id="NM_000546",
    rettype="gb",
    retmode="text"
)

with open("tp53.gb", "w") as file:
    file.write(handle.read())

handle.close()

record = SeqIO.read("tp53.gb", "genbank")

print("Organism:", record.annotations["organism"])

print("\nFirst 4 levels of Taxonomic Lineage:")
lineage = record.annotations["taxonomy"][:4]
for level in lineage:
    print(level)

print("\nMolecule Type:", record.annotations["molecule_type"])
print("Topology:", record.annotations["topology"])

references = record.annotations["references"]
print("\nTotal Number of References:", len(references))

print("\nFirst Reference Title:")
print(references[0].title)

print("\nLast Reference Title:")
print(references[-1].title)
```
